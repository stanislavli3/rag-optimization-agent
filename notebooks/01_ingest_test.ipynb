# 01 · Ingestion — loader + chunker
Load local documents, chunk at 256/512/1024 and compare chunk counts.

```python

import sys, pathlib
ROOT = pathlib.Path().resolve().parent if pathlib.Path('notebooks').exists() else pathlib.Path().resolve()
sys.path.insert(0, str(ROOT))
```

```python
from src.ingest.loader import load_documents, _Document
from src.ingest.chunker import chunk_documents

# Fake 3 docs if no sample_docs present
docs = load_documents('data/sample_docs') or [
    _Document(('The project tracks multiple agents collaborating. ' * 80),
              {'source_file': f'd{i}.txt', 'page_number': 0, 'file_type': 'txt'}) for i in range(3)
]
print(f'{len(docs)} documents loaded')
```

```python
for size in (256, 512, 1024):
    chunks = chunk_documents(docs, chunk_size=size, overlap_ratio=0.10)
    print(f'chunk_size={size:>5} -> {len(chunks)} chunks | first meta keys: {list(chunks[0].metadata)}')
```
