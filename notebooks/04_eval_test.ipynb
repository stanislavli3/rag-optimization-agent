```python
from src.evaluation.ragas_eval import evaluate_ragas, evaluate_by_difficulty, evaluate_by_question_type
from src.evaluation.ir_metrics import evaluate_ir
from src.evaluation.statistics import paired_bootstrap_test, cohens_d
import pandas as pd

results = [
    {'question':'q1','answer':'paris','retrieved_contexts':['paris is capital of france'],
     'ground_truth_answer':'paris','ground_truth_context':'paris is capital of france'},
    {'question':'q2','answer':'atp','retrieved_contexts':['mitochondria produce atp'],
     'ground_truth_answer':'atp','ground_truth_context':'mitochondria produce atp'},
]
print('RAGAS:', evaluate_ragas(results))
print('IR:', evaluate_ir([['a','b','c'],['b','a']], [['a'],['a','c']], k=3))
print('bootstrap:', paired_bootstrap_test([0.8,0.7,0.9,0.85], [0.6,0.55,0.7,0.65], n_bootstrap=500))
print('cohens_d:', cohens_d([0.8,0.7,0.9,0.85],[0.6,0.55,0.7,0.65]))

df = pd.DataFrame([
    {'question':'q1','difficulty':'easy','question_type':'simple','ground_truth_answer':'paris','ground_truth_context':'paris is capital of france'},
    {'question':'q2','difficulty':'hard','question_type':'reasoning','ground_truth_answer':'atp','ground_truth_context':'mitochondria produce atp'},
])
print('by difficulty:', evaluate_by_difficulty(results, df))
print('by question_type:', evaluate_by_question_type(results, df))
```
